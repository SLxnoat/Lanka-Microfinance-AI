```python
import pandas as pd
import numpy as np
```

```python
def dataset_generator(records=1000):
    np.random.seed(42)

    data = {
        'Applicant_ID': range(1001, 1001 + records),
        'Age': np.random.randint(20, 65, size=records),
        'Monthly_Income_LKR': np.random.randint(35000, 150000, size=records),
        'Business_Type': np.random.randint(0, 4, size=records), #Street Vendor: 0, Small Retail Shop: 1, Home-based Business: 2, Freelancer/Service Provider:3
        'Utility_Bill_Late_Days': np.random.randint(0, 25, size=records), 
        'Mobile_Reload_Consistency': np.random.uniform(0.2, 1.0, size=records),
        'Digital_Literacy_Score': np.random.uniform(0, 100, size=records),
        'Community_Group_Member': np.random.choice([0, 1], size=records, p=[0.4, 0.6]),
        'Existing_Loans': np.random.randint(0, 3, size=records),
        'Dependents': np.random.randint(0, 5, size=records),
    }

    df = pd.DataFrame(data)

    score = (df['Utility_Bill_Late_Days'] * 0.4) - (df['Mobile_Reload_Consistency'] * 15) + (df['Existing_Loans'] * 5)
    df['Loan_Default'] = (score > 5).astype(int)
    
    return df
    
df = dataset_generator(1000)
df.to_csv('data/lanka_microfinance_data.csv', index=False)
print("✅ Dataset generated successfully: lanka_microfinance_data.csv")
```
