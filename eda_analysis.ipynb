```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/lanka_microfinance_data.csv')
print("csv file load successfully.")
```

```python
plt.figure(figsize=(6,4))
sns.countplot(x='Loan_Default', data=df, palette='viridis')
plt.title('Loan Default vs Non-Default Count')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.boxplot(x='Loan_Default', y='Utility_Bill_Late_Days', data=df)
plt.title('Utility Bill Late Days vs Loan Default')
plt.show()
```

```python
plt.figure(figsize=(10, 6))
sns.violinplot(x='Loan_Default', y='Mobile_Reload_Consistency', data=df)
plt.title('Mobile Reload Consistency vs Loan Default')
plt.show()
```
